# Decision Tree

```python
X = df_train[['OverallQual', 'GrLivArea', 'GarageCars']]
y = df_train['SalePrice']
```

```python
# Cost funtion
def rmse(h, y):
    # Root mean squared error
    return sqrt(mean_squared_error(h, y))
```

```python
class Node:

    def __init__(self, x, y, idxs, min_leaf=5):
        self.x = x 
        self.y = y
        self.idxs = idxs 
        self.min_leaf = min_leaf
        self.row_count = len(idxs)
        self.col_count = x.shape[1]
        self.val = np.mean(y[idxs])
        self.score = float('inf')
        self.find_varsplit()
        
    def find_varsplit(self):
        for c in range(self.col_count): self.find_better_split(c)
        if self.is_leaf: return
        x = self.split_col
        lhs = np.nonzero(x <= self.split)[0]
        rhs = np.nonzero(x > self.split)[0]
        self.lhs = Node(self.x, self.y, self.idxs[lhs], self.min_leaf)
        self.rhs = Node(self.x, self.y, self.idxs[rhs], self.min_leaf)
        
    def find_better_split(self, var_idx):
      
        x = self.x.values[self.idxs, var_idx]

        for r in range(self.row_count):
            lhs = x <= x[r]
            rhs = x > x[r]
            if rhs.sum() < self.min_leaf or lhs.sum() < self.min_leaf: continue

            curr_score = self.find_score(lhs, rhs)
            if curr_score < self.score: 
                self.var_idx = var_idx
                self.score = curr_score
                self.split = x[r]
                
    def find_score(self, lhs, rhs):
        y = self.y[self.idxs]
        lhs_std = y[lhs].std()
        rhs_std = y[rhs].std()
        return lhs_std * lhs.sum() + rhs_std * rhs.sum()
                
    @property
    def split_col(self): return self.x.values[self.idxs,self.var_idx]
                
    @property
    def is_leaf(self): return self.score == float('inf')                

    def predict(self, x):
        return np.array([self.predict_row(xi) for xi in x])

    def predict_row(self, xi):
        if self.is_leaf: return self.val
        node = self.lhs if xi[self.var_idx] <= self.split else self.rhs
        return node.predict_row(xi)
```

```python
class DecisionTreeRegressor:
  
    def fit(self, X, y, min_leaf = 5):
        self.dtree = Node(X, y, np.array(np.arange(len(y))), min_leaf)
        return self
  
    def predict(self, X):
        return self.dtree.predict(X.values)
```

```python
regressor = DecisionTreeRegressor().fit(X, y)
preds = regressor.predict(X)
```

```python
# R2 score
metrics.r2_score(y, preds)
```

```python
# Root MSE
rmse(preds, y)
```

## Fronteira de decisao 2D
Para ver como a arvore corta o espaco em retangulos, projetamos em duas features (OverallQual e GrLivArea), transformamos o preco em alvo binario (acima vs abaixo da mediana) e pintamos um meshgrid 2D usando a arvore feita a mao.

```python
import matplotlib.pyplot as plt

features = ['OverallQual', 'GrLivArea']
X_plot = df_train[features]
y_binary = (df_train['SalePrice'] > df_train['SalePrice'].median()).astype(int)

# Ajusta uma arvore simples com duas variaveis para conseguir visualizar o plano 2D
tree_2d = DecisionTreeRegressor().fit(X_plot, y_binary)

x_min, x_max = X_plot[features[0]].min() - 1, X_plot[features[0]].max() + 1
y_min, y_max = X_plot[features[1]].min() - 50, X_plot[features[1]].max() + 50
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)

grid = pd.DataFrame(np.c_[xx.ravel(), yy.ravel()], columns=features)
Z = tree_2d.predict(grid).reshape(xx.shape)
Z_binary = (Z >= 0.5).astype(int)

plt.figure(figsize=(10, 6))
cmap = plt.cm.coolwarm
plt.contourf(xx, yy, Z_binary, levels=[-0.1, 0.5, 1.1], alpha=0.35, cmap=cmap)
plt.scatter(X_plot[features[0]], X_plot[features[1]], c=y_binary, cmap=cmap, edgecolor='k', s=35)
plt.xlabel(features[0])
plt.ylabel(features[1])
plt.title('Approximate decision boundary (manually implemented tree)')
plt.show()
```
